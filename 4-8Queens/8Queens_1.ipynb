# N-Queens Genetic Algorithm
This notebook demonstrates solving the N-Queens problem using a genetic algorithm.

```python
# Imports and parameters
import random

N = 8
POP_SIZE = 200
MUTATION_RATE = 0.15
ELITE_SIZE = 10
MAX_GENERATIONS = 5000
```

## Fitness Function
Counts the number of diagonal conflicts in a board.

```python
# Fitness function
def fitness(board):
    conflicts = 0
    for i in range(N):
        for j in range(i + 1, N):
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts
```

## Individual Creation
Creates a random board (individual) for the population.

```python
# Individual creation
def create_individual():
    board = list(range(N))
    random.shuffle(board)
    return board
```

## Selection, Crossover, Mutation
Functions for tournament selection, crossover, and mutation.

```python
# Selection, crossover, mutation
def tournament_selection(population):
    tournament = random.sample(population, 5)
    tournament.sort(key=fitness)
    return tournament[0]

def crossover(p1, p2):
    start, end = sorted(random.sample(range(N), 2))
    child = [None] * N
    child[start:end] = p1[start:end]

    pointer = 0
    for gene in p2:
        if gene not in child:
            while child[pointer] is not None:
                pointer += 1
            child[pointer] = gene

    return child

def mutate(board):
    if random.random() < MUTATION_RATE:
        i, j = random.sample(range(N), 2)
        board[i], board[j] = board[j], board[i]
```

## Genetic Algorithm and Execution
Runs the genetic algorithm and prints the solution.

```python
# Genetic algorithm and execution
def genetic_algorithm():
    population = [create_individual() for _ in range(POP_SIZE)]

    for generation in range(MAX_GENERATIONS):
        population.sort(key=fitness)

        if fitness(population[0]) == 0:
            return population[0], generation

        next_generation = population[:ELITE_SIZE]

        while len(next_generation) < POP_SIZE:
            parent1 = tournament_selection(population)
            parent2 = tournament_selection(population)
            child = crossover(parent1, parent2)
            mutate(child)
            next_generation.append(child)

        population = next_generation

    return population[0], MAX_GENERATIONS

solution, generation = genetic_algorithm()
print("Solution:", solution)
print("Generation:", generation)
print("Conflicts:", fitness(solution))
```

```python
def print_board(board):
    N = len(board)
    for row in range(N):
        line = ['.'] * N
        line[board[row]] = 'Q'
        print(' '.join(line))
    print()


print_board(solution)
```
